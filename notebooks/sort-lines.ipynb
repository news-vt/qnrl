```python
### Process QnRL.

def process_csv_hyp_c51(fpath: Path):

    # Read CSV into dataframe.
    df = pd.read_csv(fpath)

    # Remove duplicate entries for seed group hash ID.
    df = df.drop_duplicates(subset=["seed_group_hash"], keep="first")
    
    # Sort seed group hash ID.
    df = df.sort_values(by=["seed_group_hash"])

    rows = []
    for row in df.itertuples(index=False, name=None):
        row = row[2:] # Cut name and sweep ID.
        row = list(row)
        
        # Get first chunk of seed group hash ID and convert to latex `\texttt{}`.
        row[0] = f"\\texttt{{{row[0].split('-')[0]}}}"
        
        for i in range(1, 7):
            row[i] = f"${row[i]}$"
        
        rows.append(row)

    s = "\n".join(" & ".join(row)for row in rows)
    return s

def process_csv_hyp_qnrl(fpath: Path):

    # Read CSV into dataframe.
    df = pd.read_csv(fpath)

    # Remove duplicate entries for seed group hash ID.
    df = df.drop_duplicates(subset=["seed_group_hash"], keep="first")
    
    # Sort seed group hash ID.
    df = df.sort_values(by=["seed_group_hash"])

    rows = []
    for row in df.itertuples(index=False, name=None):
        row = row[2:] # Cut name and sweep ID.
        row = list(row)
        
        # Get first chunk of seed group hash ID and convert to latex `\texttt{}`.
        row[0] = f"\\texttt{{{row[0].split('-')[0]}}}"
        
        for i in range(1, 9):
            row[i] = f"${row[i]}$"
        
        for i in range(9, 11):
            row[i] = f"${row[i].replace('"', "")}$"
        
        row[11] = f"\\texttt{{{row[11].split('-')[0].replace('circ', 'circular')}}}"
        
        row[12] = f"\\texttt{{{row[12]}}}"
        
        rows.append(row)

    s = "\n".join("& " + " & ".join(row) + " \\\\" for row in rows)
    return s


def process_csv_eval_all(fpath: Path):
    # Read CSV into dataframe.
    df = pd.read_csv(fpath)

    # Remove duplicate entries for seed group hash ID.
    df = df.drop_duplicates(subset=["seed_group_hash"], keep="first")
    
    # Sort seed group hash ID.
    df = df.sort_values(by=["seed_group_hash"])
    
    rows = []
    for row in df.itertuples(index=False, name=None):
        row = row[2:] # Cut name and sweep ID.
        row = list(row)
        
        # Get first chunk of seed group hash ID and convert to latex `\texttt{}`.
        row[0] = f"\\texttt{{{row[0].split('-')[0]}}}"
        
        # Formulate rest of the columns as numbers within math `$$`.
        for i in range(1, 3):
            if isinstance(row[i], int):
                row[i] = f"${row[i]}$"
            else:
                row[i] = f"${row[i]:.04f}$"
        
        rows.append(row)
    
    s = "\n".join(" & ".join(row) for row in rows)
    return s


fpaths = [
    # Path("~/Downloads/c51_CartPole_hyp.csv"),
    # Path("~/Downloads/c51_CliffWalking_hyp.csv"),
    # Path("~/Downloads/c51_FrozenLake_hyp.csv"),
    # Path("~/Downloads/c51_Acrobot_hyp.csv"),
    #
    Path("~/Downloads/c51_CartPole_eval.csv"),
    Path("~/Downloads/c51_CliffWalking_eval.csv"),
    Path("~/Downloads/c51_FrozenLake_eval.csv"),
    Path("~/Downloads/c51_Acrobot_eval.csv"),
    # # 
    # Path("~/Downloads/qnrl_CartPole_hyp.csv"),
    # Path("~/Downloads/qnrl_CliffWalking_hyp.csv"),
    # Path("~/Downloads/qnrl_FrozenLake_hyp.csv"),
    # Path("~/Downloads/qnrl_Acrobot_hyp.csv"),
    # #
    # Path("~/Downloads/qnrl_CartPole_eval.csv"),
    # Path("~/Downloads/qnrl_CliffWalking_eval.csv"),
    # Path("~/Downloads/qnrl_FrozenLake_eval.csv"),
    # Path("~/Downloads/qnrl_Acrobot_eval.csv"),
]



for fpath in fpaths:

    if 'c51' in fpath.stem:
        if '_hyp' in fpath.stem:
            s = process_csv_hyp_c51(fpath)
        if '_eval' in fpath.stem:
            s = process_csv_eval_all(fpath)
    elif 'qnrl' in fpath.stem:
        if '_hyp' in fpath.stem:
            s = process_csv_hyp_qnrl(fpath)
        if '_eval' in fpath.stem:
            s = process_csv_eval_all(fpath)
    
    print(fpath.stem)
    print(s)



# # fpath = Path("~/Downloads/qnrl_CartPole_eval.csv")
# # fpath = Path("~/Downloads/qnrl_CliffWalking_eval.csv")
# # fpath = Path("~/Downloads/qnrl_FrozenLake_eval.csv")
# fpath = Path("~/Downloads/qnrl_Acrobot_eval.csv")
# s = process_csv_eval_all(fpath)
```
